```python
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]
print("Uploaded file:", filename)
```

```python
import subprocess

cmd = f'ffprobe -v error -select_streams v:0 -show_entries stream=codec_name,pix_fmt,width,height,r_frame_rate -of default=noprint_wrappers=1 "{filename}"'
print(subprocess.check_output(cmd, shell=True).decode())
```

```python
output = "whatsapp_fixed.mp4"

!ffmpeg -i "$filename" \
-vcodec libx264 \
-profile:v main \
-pix_fmt yuv420p \
-r 30 \
-b:v 4M \
-acodec aac \
-movflags +faststart \
"$output"
```
